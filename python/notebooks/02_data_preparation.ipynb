## 3. Define Observation Cutoff

### 3.1. Observation Cutoff

#### Available Evidence
The latest recorded subscription creation and cancellation dates are September 8th, 2023. However, the official data collection cutoff has not been confirmed.
#### Proposed Business Rule
September 8th, 2023, will be used as the provisional observation cutoff for this analysis. Monthly performance reporting will cover Septermber 2022, through August 2023, excluding September 2023 from full-month comparisons.
#### Analytical Rationale
September 2023 contains only eight days of observed activity under the assumed cutoff. Including it alongside complete months could make subscription growth and cancellation volumes appear artificially low, leading to misleading performance comparisons.
#### Limitation
The observation cutoff is inferred from the latest recorded dates rather than confirmed source-system documentation. Therefore, conclusions about subscription activity and retention will depend on this assumption.

### 3.2. Missing Cancellation Dates

#### Available Evidence
Of the 3,069 subscription records, 1,065 have no recorded cancellation date.
#### Proposed Business Rule
Subscriptions without a cancellation date will be treated as active as of the provisional observation cutoff, provided their creation date is on or before the cutoff.

Subscription with a cancellation date on or before the cutoff will be classified as canceled.
#### Analytical Rationale
A missing cancellation date indicates that no cancellation was recorded in the available data. Treating such records as active allows subscription status and retention to be estimated consistently.
#### Limitation
This classification assumes cancellation records are complete through the cutoff. It does not confirm continuous payment, uninterrupted service access, or the customer's actual subscription status beyond the observation period.

### 3.3. Payment Status Preperation

The `was_subscription_paid` field indicates whether a subscription was recorded as paid. Payment status and cancellation status will be treated independently. An unpaid subscription will no automatically be classified as canceled, since the dataset does not establish a direct relationship between payment failure and subscription termination.
